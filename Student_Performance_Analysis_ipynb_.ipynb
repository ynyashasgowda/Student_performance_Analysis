```python


# MINI PROJECT: STUDENT ACADEMIC PERFORMANCE ANALYSIS

# 1. Create Dataset
students <- c("Asha", "Bharath", "Chitra", "Deepak",
              "Esha", "Farhan", "Gita", "Harish",
              "Isha", "Jeevan")

Maths <- c(85,76,90,65,88,72,95,80,68,85)
Science <- c(78,82,91,70,85,75,92,78,72,80)
English <- c(92,75,89,68,90,70,96,82,75,88)
Computer <- c(88,80,94,72,86,74,90,85,70,82)
Social <- c(90,78,92,66,89,73,94,79,74,86)

data <- data.frame(students, Maths, Science,
                   English, Computer, Social)

cat("STUDENT DATASET\n")
print(data)

# 2. Calculate Total Marks
data$Total <- rowSums(data[2:6])

cat("\nTOTAL MARKS\n")
print(data[, c("students", "Total")])

# 3. Define Mode Function
mode_value <- function(x) {
  t <- table(x)
  m <- as.numeric(names(t)[t == max(t)])

  if(length(m) == length(unique(x))) {
    return("No mode")
  }

  return(paste(m, collapse = ", "))
}

# 4. Statistical Analysis
subjects <- data[2:6]

cat("\nMEAN\n")
print(round(sapply(subjects, mean), 2))

cat("\nMEDIAN\n")
print(sapply(subjects, median))

cat("\nMODE\n")
print(sapply(subjects, mode_value))

cat("\nVARIANCE\n")
print(round(sapply(subjects, var), 2))

cat("\nSTANDARD DEVIATION\n")
print(round(sapply(subjects, sd), 2))

# 5. Highest and Lowest Scoring Students
cat("\nHIGHEST SCORING STUDENT\n")
print(data[which.max(data$Total), c("students", "Total")])

cat("\nLOWEST SCORING STUDENT\n")
print(data[which.min(data$Total), c("students", "Total")])

# 6. Graph 1 - Total Marks of Students
barplot(data$Total,
        names.arg = data$students,
        col = "skyblue",
        main = "Total Marks of Students",
        xlab = "Students",
        ylab = "Total Marks",
        las = 2)

# 7. Graph 2 - Average Marks by Subject
barplot(colMeans(subjects),
        col = "lightgreen",
        main = "Average Marks in Each Subject",
        xlab = "Subjects",
        ylab = "Average Marks")

# 8. Graph 3 - Box Plot
boxplot(subjects,
        col = "lightpink",
        main = "Marks Distribution by Subject",
        ylab = "Marks")
```
